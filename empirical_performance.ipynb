```python
def make_results_table(X, y, k, m_values):
    knn_s = score_knn(X, k=k)
    lof_s = score_lof(X, k=k)
    loci_s = score_loci(X, k=k)
    inflo_s = score_inflo(X, k=k)
    cof_s = score_cof(X, k=k)
    ccof_s = score_ccof(X, k=k)
    
    out = []
    for m in m_values:
        a = get_metrics(knn_s, y, m)
        b = get_metrics(lof_s, y, m)
        c = get_metrics(loci_s, y, m)
        d = get_metrics(inflo_s, y, m)
        e = get_metrics(cof_s, y, m)
        f = get_metrics(ccof_s, y, m)
        
        out.append({
            "m": m,
            "knn_mt": a["mt"], "knn_Re": a["Re"], "knn_RP": a["RP"],
            "lof_mt": b["mt"], "lof_Re": b["Re"], "lof_RP": b["RP"],
            "loci_mt": c["mt"], "loci_Re": c["Re"], "loci_RP": c["RP"],
            "inflo_mt": d["mt"], "inflo_Re": d["Re"], "inflo_RP": d["RP"],
            "cof_mt": e["mt"], "cof_Re": e["Re"], "cof_RP": e["RP"],
            "ccof_mt": f["mt"], "ccof_Re": f["Re"], "ccof_RP": f["RP"]
        })
        
    return pd.DataFrame(out)


def print_latex_rows(df):
    for _, row in df.iterrows():
        print(
            f"{int(row['m'])} & "
            f"{int(row['knn_mt'])} & {row['knn_Re']:.3f} & {row['knn_RP']:.3f} & "
            f"{int(row['lof_mt'])} & {row['lof_Re']:.3f} & {row['lof_RP']:.3f} & "
            f"{int(row['loci_mt'])} & {row['loci_Re']:.3f} & {row['loci_RP']:.3f} & "
            f"{int(row['inflo_mt'])} & {row['inflo_Re']:.3f} & {row['inflo_RP']:.3f} & "
            f"{int(row['cof_mt'])} & {row['cof_Re']:.3f} & {row['cof_RP']:.3f} & "
            f"{int(row['ccof_mt'])} & {row['ccof_Re']:.3f} & {row['ccof_RP']:.3f} \\\\\n\\hline"
        )

k_values = [10, 20, 30, 40, 50]
m_values = [5, 10, 15, 20, 30]

if __name__ == "__main__":
    # Load dataset (change to dat1.txt, dat2.txt, or dat3.txt)
    #file_path = "/Users/smahmood1/Library/CloudStorage/Dropbox/MyPapers/Anomaly/dat1.txt"
    file_path = "/Users/smahmood1/Library/CloudStorage/Dropbox/MyPapers/Anomaly/dat2.txt"
    dat = pd.read_csv(file_path, sep=r'\s+')
    
    feature_cols = [c for c in dat.columns if c not in ['label']]
    X = dat[feature_cols].values
    y = np.where(dat['label'] == 'o', 1, 0)
    
    # Run evaluations across different k parameters
    for k_val in k_values:
        print(f"\n %LaTeX Table Output for k = {k_val} %")
        print(rf"\multicolumn{{1}}{{|c|}}{{${{k}}={k_val}$}} & \multicolumn{{3}}{{|c|}}{{K-NN}} & \multicolumn{{3}}{{|c|}}{{LOF}}")
        print(r"    & \multicolumn{3}{|c|}{LOCI} & \multicolumn{3}{|c|}{INFLO} & \multicolumn{3}{|c|}{COF} & \multicolumn{3}{|c|}{CCOF}\\")
        print(r"    \hline")
        print(r"    $m$ & $m_t$ & Re & RP & $m_t$ & Re & RP & $m_t$ & Re & RP & $m_t$ & Re & RP & $m_t$ & Re & RP & $m_t$ & Re & RP \\")
        print(r"\hline ")

        tab = make_results_table(X, y, k=k_val, m_values=m_values)
        print_latex_rows(tab)
```

```python
import numpy as np
from sklearn.metrics import roc_auc_score, precision_recall_curve, auc
from scipy.spatial.distance import cdist

# Import anomaly scoring functions from your custom module
from anomaly_scores import (
    score_knn,
    score_lof,
    score_cof,
    score_loci,
    score_inflo,
    score_ccof
)


def compute_auc_pr(y_true, y_scores):
    """Calculates Area Under the Precision-Recall Curve (AUC-PR)."""
    precision, recall, _ = precision_recall_curve(y_true, y_scores)
    return auc(recall, precision)


# =====================================================================
# 1. SYNTHETIC TOPOLOGY GENERATORS
# =====================================================================

def generate_global_local_boundary(seed=42):
    """Topology 1: Main Gaussian cluster with boundary and global outliers."""
    np.random.seed(seed)
    # Main normal cluster
    X_normal = np.random.normal(loc=0.0, scale=1.0, size=(250, 2))
    # Global anomalies
    X_global = np.random.uniform(low=-8.0, high=8.0, size=(10, 2))
    # Filter to keep true global anomalies far from origin
    dist_from_center = np.linalg.norm(X_global, axis=1)
    X_global = X_global[dist_from_center > 5.0]
    
    X = np.vstack([X_normal, X_global])
    y = np.array([0] * len(X_normal) + [1] * len(X_global))
    return X, y


def generate_non_convex_manifolds(seed=42):
    """Topology 2: Non-convex spiral / half-moon geometry with anomalies."""
    np.random.seed(seed)
    n_samples = 150
    
    # Half-moon 1
    t1 = np.linspace(0, np.pi, n_samples)
    x1 = np.cos(t1) + np.random.normal(0, 0.08, n_samples)
    y1 = np.sin(t1) + np.random.normal(0, 0.08, n_samples)
    
    # Half-moon 2
    t2 = np.linspace(0, np.pi, n_samples)
    x2 = 1.0 - np.cos(t2) + np.random.normal(0, 0.08, n_samples)
    y2 = 0.5 - np.sin(t2) + np.random.normal(0, 0.08, n_samples)
    
    X_normal = np.vstack([np.column_stack([x1, y1]), np.column_stack([x2, y2])])
    
    # Anomalies placed near center / between moons
    X_anomalies = np.array([
        [0.5, 0.2], [0.5, -0.2], [-0.5, 0.5], 
        [1.5, 0.0], [0.0, 1.2], [1.0, -0.8]
    ])
    
    X = np.vstack([X_normal, X_anomalies])
    y = np.array([0] * len(X_normal) + [1] * len(X_anomalies))
    return X, y


def generate_multi_density_clusters(seed=42):
    """Topology 3: High-density and low-density clusters with anomalies."""
    np.random.seed(seed)
    # Dense cluster
    X_dense = np.random.normal(loc=[-4.0, 0.0], scale=0.4, size=(200, 2))
    # Sparse cluster
    X_sparse = np.random.normal(loc=[4.0, 0.0], scale=1.8, size=(100, 2))
    # Isolated anomalies
    X_anomalies = np.array([
        [0.0, 0.0], [0.0, 3.0], [0.0, -3.0], 
        [-8.0, 0.0], [9.0, 0.0], [4.0, 6.0]
    ])
    
    X = np.vstack([X_dense, X_sparse, X_anomalies])
    y = np.array([0] * (len(X_dense) + len(X_sparse)) + [1] * len(X_anomalies))
    return X, y


def generate_micro_clusters(seed=42):
    """Topology 4: Small collective anomaly / micro-cluster (S <= k)."""
    np.random.seed(seed)
    # Primary cluster (N = 300)
    X_main = np.random.normal(loc=0.0, scale=1.0, size=(300, 2))
    # Micro-cluster anomaly (S = 4)
    X_micro = np.random.normal(loc=[12.0, 12.0], scale=0.1, size=(4, 2))
    
    X = np.vstack([X_main, X_micro])
    y = np.array([0] * len(X_main) + [1] * len(X_micro))
    return X, y


# =====================================================================
# 2. EVALUATION PIPELINE & LATEX TABLE BUILDER
# =====================================================================

def run_synthetic_auc_evaluation(k=10):
    topologies = {
        "Global & Local Boundary": generate_global_local_boundary(),
        "Non-Convex Manifolds": generate_non_convex_manifolds(),
        "Multi-Density Clusters": generate_multi_density_clusters(),
        "Small Micro-Clusters ($S \\le k$)": generate_micro_clusters()
    }
    
    methods = {
        "$k$-NN": score_knn,
        "LOF": score_lof,
        "COF": score_cof,
        "INFLO": score_inflo,
        "LOCI": score_loci,
        "CCOF": score_ccof
    }
    
    results = {}
    
    for top_name, (X, y) in topologies.items():
        results[top_name] = {"AUC-ROC": {}, "AUC-PR": {}}
        
        for method_name, score_fn in methods.items():
            try:
                # Compute anomaly scores
                scores = score_fn(X, k=k)
                
                # Replace any NaN/Inf with 0
                scores = np.nan_to_num(scores, nan=0.0, posinf=0.0, neginf=0.0)
                
                auc_roc = roc_auc_score(y, scores)
                auc_pr = compute_auc_pr(y, scores)
                
                results[top_name]["AUC-ROC"][method_name] = auc_roc
                results[top_name]["AUC-PR"][method_name] = auc_pr
            except Exception as e:
                results[top_name]["AUC-ROC"][method_name] = 0.0
                results[top_name]["AUC-PR"][method_name] = 0.0
                print(f"Error executing {method_name} on {top_name}: {e}")

    # Build LaTeX Table
    latex_output = []
    latex_output.append("\\begin{table*}[t]")
    latex_output.append("\\centering")
    latex_output.append("\\caption{Empirical performance comparison across synthetic benchmark topologies measured by AUC-ROC and AUC-PR. Bold values indicate superior performance.}")
    latex_output.append("\\label{tab:synthetic_auc}")
    latex_output.append("\\scalebox{0.8}{")
    latex_output.append("\\begin{tabular}{llcccccc}")
    latex_output.append("\\hline")
    latex_output.append("\\textbf{Topology} & \\textbf{Metric} & \\textbf{$k$-NN} & \\textbf{LOF} & \\textbf{COF} & \\textbf{INFLO} & \\textbf{LOCI} & \\textbf{CCOF} \\\\ \\hline")
    
    for top_idx, (top_name, metrics) in enumerate(results.items()):
        for metric_name in ["AUC-ROC", "AUC-PR"]:
            row_scores = metrics[metric_name]
            max_val = max(row_scores.values())
            
            row_str = []
            if metric_name == "AUC-ROC":
                row_str.append(f"{top_name} & {metric_name}")
            else:
                row_str.append(f" & {metric_name}")
                
            for method_name in methods.keys():
                val = row_scores[method_name]
                val_fmt = f"{val:.3f}"
                
                # Bold max value (or within floating-point epsilon)
                if abs(val - max_val) < 1e-4:
                    val_fmt = f"\\textbf{{{val_fmt}}}"
                    
                row_str.append(val_fmt)
                
            line = " & ".join(row_str) + " \\\\"
            if metric_name == "AUC-PR":
                line += " \\hline"
            latex_output.append(line)
            
    latex_output.append("\\end{tabular}%")
    latex_output.append("}")
    latex_output.append("\\end{table*}")
    
    print("\n".join(latex_output))


if __name__ == "__main__":
    run_synthetic_auc_evaluation(k=10)
```
